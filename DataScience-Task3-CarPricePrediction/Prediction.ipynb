## 1. Linear Regression Model

Train a Linear Regression model to predict car selling prices.

## 5. Model Evaluation and Comparison

We trained and evaluated three regression models for car price prediction:

1. **Linear Regression** – learns a linear relationship between car features and selling price.
2. **Decision Tree Regressor** – predicts prices using decision-based splits.
3. **Gradient Boosting Regressor** – builds trees sequentially, with each new tree attempting to correct the remaining prediction errors.

### Evaluation Metrics

We used the following regression metrics to compare the models:

- **MAE (Mean Absolute Error):** Measures the average absolute difference between predicted and actual prices.
- **MSE (Mean Squared Error):** Calculates the average squared prediction error and penalizes larger errors more heavily.
- **RMSE (Root Mean Squared Error):** Measures prediction error in the same units as the selling price.
- **R² Score:** Measures how much variation in selling prices the model explains relative to a mean-price baseline.

### Initial Results

The Decision Tree Regressor achieved the best performance on our current test set, with an MAE of 1.0105 and an R² score of approximately 0.8449.

Gradient Boosting and Linear Regression performed less well on this particular split. We will use five-fold cross-validation on the training data to assess the Decision Tree's performance more reliably before finalizing the model.

**Cross Validation**

**Cross Validation for LinearRegression , DecisionTreeRegressor and GradientBoost**

```python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import GradientBoostingRegressor

models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42)
}

for name, model in models.items():
    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=5,
        scoring="r2"
    )

    print(name)
    print("Fold scores:", scores.round(4))
    print("Mean R2:", round(scores.mean(), 4))
    print("-" * 40)
```

** Comparasion Table for MAE,RSME and all other metrices**

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pandas as pd

results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Gradient Boosting"
    ],
    "MAE": [
        mean_absolute_error(y_test, y_pred),
        mean_absolute_error(y_test, y_prediction),
        mean_absolute_error(y_test, gb_prediction)
    ],
    "RMSE": [
        mean_squared_error(y_test, y_pred) ** 0.5,
        mean_squared_error(y_test, y_prediction) ** 0.5,
        mean_squared_error(y_test, gb_prediction) ** 0.5
    ],
    "Test R2": [
        r2_score(y_test, y_pred),
        r2_score(y_test, y_prediction),
        r2_score(y_test, gb_prediction)
    ],
    "Mean CV R2": [
        0.7508,
        0.8683,
        0.8675
    ]
})

results.round(4)
```

### 7. Model Comparison and Selection

Three regression models were evaluated for the Car Price Prediction project: Linear Regression, Decision Tree Regressor, and Gradient Boosting Regressor.

The models were compared using Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), test-set R² score, and the mean R² score obtained from 5-fold cross-validation.

**Decision Tree Regressor performed best overall**, achieving an MAE of 1.0105, RMSE of 1.9992, test R² of 0.8449, and mean cross-validation R² of 0.8683.

Based on these results, Decision Tree Regressor was selected as the current best-performing model. Its performance should still be evaluated for generalization and potential overfitting before deployment.
